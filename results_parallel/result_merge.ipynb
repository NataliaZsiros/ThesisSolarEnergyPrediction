```python
import pandas as pd

df_xgb = pd.read_csv('parallel_XGBoost_metrics.csv')
df_svr = pd.read_csv('parallel_svr_fitting_metrics.csv')
df_lstm = pd.read_csv('parallel_LSTM_metrics.csv')
df_gpr = pd.read_csv('parallel_gpr_metrics.csv')
df_linreg = pd.read_csv('parallel_linear_regression_metrics.csv')
persistence_df = pd.read_csv('persistence_rmse.csv')
```

```python
final_df = pd.concat([df_xgb, df_svr, df_lstm, df_gpr, df_linreg], ignore_index=True)
```

```python
final_df
```

```python
rmse_value = persistence_df.loc[0, 'RMSE']
rmse_value
```

```python
final_df['skill_score'] = 1 - (final_df['RMSE'] / rmse_value)
final_df
```

```python
training_time_table = final_df[['Model','Training Time (s)']].sort_values(by=['Training Time (s)'])
prediction_time_table = final_df[['Model','Prediction Time (s)']].sort_values(by=['Prediction Time (s)'])
mae_table = final_df[['Model','MAE']].sort_values(by=['MAE'])
r_squared_table = final_df[['Model','R-squared']].sort_values(by=['R-squared'])
rmse_table = final_df[['Model','RMSE']].sort_values(by=['RMSE'])
mse_table = final_df[['Model','MSE']].sort_values(by=['MSE'])
```

```python
training_time_table
```

```python
ranking_df = pd.DataFrame()
ranking_df["Model"] = final_df["Model"]

for col in ["Training Time (s)", "Prediction Time (s)", "MAE", "RMSE", "MSE"]:
    ranking_df[col] = final_df[col].rank(method="min", ascending=True).astype(int)

ranking_df["R-squared"] = final_df["R-squared"].rank(method="min", ascending=False).astype(int) 
ranking_df["skill_score"] = final_df["skill_score"].rank(method="min", ascending=False).astype(int) 

# Display the ranking table
ranking_df
```

```python
ranking_df["Row Average"] = ranking_df.iloc[:, 1:].mean(axis=1)
```
